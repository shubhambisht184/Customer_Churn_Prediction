```python
# Step 1 : Import Libraries
import pandas as pd
import numpy as np
```

```python
# Step 2: Load Datasets
df = pd.read_excel("Churn_Unclean_Project.xlsx")
```

```python
# Step 4 : Check Missing Values 

df.isnull().sum()
```

```python
# step 5 : check duplicates records 

df.duplicated().sum()
```

```python
# Remove Duplicate 

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.shape
```

```python
# Step 6 : Replace Dirty Values
df.replace(["N/A","NULL",""," "],np.nan,inplace=True)
```

```python
df.isnull().sum()
```

```python
df.head()
```

```python
# Step 7 : Remove Extra Spaces 

df=df.apply(lambda x:x.str.strip() if x.dtype=="object" else x)
```

```python
# Value count 
df["Subscription_Type"].value_counts()
```

```python
# Step 8 : Proper Case
df["Customer_Name"]=df["Customer_Name"].str.title()
df["State"]=df["State"].str.title()
df["City"]=df["City"].str.title()
df["Subscription_Type"]=df["Subscription_Type"].str.title()
df["Contract_Type"]=df["Contract_Type"].str.title()
df['Churn']=df['Churn'].str.title()
df["Gender"]=df["Gender"].str.title()
# df=df.apply(lambda x:x.str.title() if x.dtype=="object" else x)
```

```python
df["Subscription_Type"].value_counts()
```

```python
# Step 9 : Standardize Churn Column

df['Churn'].value_counts()
```

```python
# Step 10 : Convert Numeric Columns
cols=["Age","Tenure_Months","Monthly_Charges","Total_Charges"]
for c in cols:
    df[c]=pd.to_numeric(df[c],errors="coerce")
```

```python
df.describe()
```

```python
# Step 11 : Remove Invalid Age
    
df=df[(df["Age"]>=18)&(df["Age"]<=100)]
```

```python
# Step 12 : Remove Negative Charges

df=df[df["Monthly_Charges"]>=0]
df=df[df["Total_Charges"]>=0]
```

```python
df.shape
```

```python
# Step 13: Convert Date

df["Last_Interaction_Date"]=pd.to_datetime(df["Last_Interaction_Date"],errors="coerce")
```

```python
# Step 14 : Fill Missing Values

df["Tech_Support"]=df["Tech_Support"].fillna("No")
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")
df["Monthly_Charges"]=df["Monthly_Charges"].fillna(df['Monthly_Charges'].mean())
df["Tenure_Months"]=df["Tenure_Months"].fillna(df['Tenure_Months'].mean())
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
# Step 15 : Feature Engineering

# Customer Lifetime Value

df["Customer_Value"]=df["Monthly_Charges"]*df["Tenure_Months"]
```

```python
df.head()
```

```python
# Monthly Revenue

df["Monthly_Revenue"]=df["Monthly_Charges"]
```

```python
# Tenure Group "0-12","13-24","25-48","49-72"
bins = [0, 12, 24, 48, 72]
labels = ["0-12 month", "13-24 month", "25-48 month", "49-72 month"]
df["Tenure_Group"] = pd.cut(df["Tenure_Months"], bins=bins, labels=labels)

```

```python
df.head()
```

```python
# Senior Citizen Flag
df["Senior_Flag"]=np.where(df["Age"]>=60,"Senior","Adult")
```

```python
df.head()
```

```python
# Churn Flag
df["Churn_Flag"]=df["Churn"].map({"Yes":1,"No":0})
```

```python
df.head()
```

```python
# Step 16 : Final Check
df.info()
```
